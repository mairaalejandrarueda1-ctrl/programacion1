```python
import menu
import entrada
import operaciones_matrices as ops

def ejecutar():
    while True:
        opc = menu.mostrar_menu()
        if opc == 5:
            print("Saliendo...")
            break
        print("\n--- Datos Matriz A ---")
        A = entrada.ingresar_matriz()
        print("\n--- Datos Matriz B ---")
        B = entrada.ingresar_matriz()
        res = None
        if opc == 1: res = ops.sumar_matrices(A, B)
        elif opc == 2: res = ops.multiplicar_matrices(A, B)
        elif opc == 3: res = ops.hadamard_matrices(A, B)
        elif opc == 4: res = ops.kronecker(A, B)
        if res:
            print("\nResultado:")
            entrada.mostrar_matriz(res)
        else:
            print("\nError: Dimensiones no compatibles.")

if __name__ == "__main__":
    ejecutar()
```
