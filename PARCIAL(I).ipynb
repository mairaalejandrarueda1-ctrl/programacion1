**Código 2:** Cree una función cuya entrada sea un objeto tipo str, el cual cuente:
* Vocales mayúsculas y minúsculas,
* Consonantes mayúsculas y minúsculas,
 números y
* otros caracteres (como espacios, números o sı́mbolos)

La salida de la función es un diccionario con la cantidad total de vocales, consonantes y otros
caracteres encontrados en el texto.
Ejecute la función con una cadena ingresada por el usuario.

```python
def contador_vocales1(palabra):
  letras = "aeioubcdfghjklmnñpqrstvwxyz"
  letras1 = letras.upper()
  numeros = "0123456789"
  espacio = " "

  contador = 0
  contador1 = 0
  contador2 = 0
  contador3 = 0
  contador4 = 0

  for i in palabra:
    if i in letras:
      contador += 1
    elif i in letras1:
      contador1 += 1
    elif i in numeros:
        contador2 += 1
    elif i in espacio:
      contador3 += 1
    else:
      contador4 += 1

  return({"Cantidad de letras minusculas:": contador,
           "Cantidad de letra mayusculas:": contador1,
           "Cantidad de números:": contador2,
           "Cantidad de espacios": contador3,
           "Cantidad de otros caracteres": contador4})
```

```python
contador_vocales1(" Maira2221977@Correo.uis.Edu.CO ")
```

```python
palabrea = input("Ingrese una palabra: ")
contador_vocales1(palabrea)
```
