```python
import re

try:
    nombre = input("Ingrese su nombre y apellido sin espacios: ").lower()
    edad = input("Ingrese su edad: ")
    # Validar nombre
    if not re.fullmatch(r"[a-zA-Z]+", nombre):
        raise ValueError("El nombre solo debe contener letras")
    # Validar edad
    if not re.fullmatch(r"\d+", edad):
        raise ValueError("La edad debe contener solo números")
    # Crear contraseña
    contrasena = nombre[:3] + edad + "@2026"+
    print("Contraseña generada:")
    print(contrasena)
except ValueError as e:
    print("Error:", e)
```

```python
import re

try:
    nombre = input("Ingrese su nombre: ").lower()
    apellido = input("Ingrese su apellido: ")
    CC = input("Ingrese su numero de CC o TI: ")
    # Validar nombre
    if not re.fullmatch(r"[a-zA-Z]+", nombre):
        raise ValueError("El nombre inválido")
    # Validar apellido
    if not re.fullmatch(r"[a-zA-Z]+", apellido):
        raise ValueError("Apellido inválido")
    if not re.fullmatch(r"\d+", CC):
        raise ValueError("CC o TI inválido")
    # Correo
    correo = nombre + apellido + CC[:3] + "@uis.edu.co"
    print("Correo generada:")
    print(correo)
except ValueError as e:
    print("Error:", e)
```

```python
import re

try:
    nombre_completo = input("Ingrese su nombre completo: ").lower()

    # Validar que solo haya letras y espacios
    if not re.fullmatch(r"[a-zA-Záéíóúñ\s]+", nombre_completo):
        raise ValueError("Solo se permiten letras y espacios")

    partes = nombre_completo.split()

    # Validar mínimo nombre y apellido
    if len(partes) < 2:
        raise ValueError("Debe ingresar al menos nombre y apellido")

    # Caso: un nombre y un apellido
    if len(partes) == 2:
        primer_nombre = partes[0]
        primer_apellido = partes[1]

        correo = (
            primer_nombre[0]
            + primer_apellido
            + "@gmail.com"
        )

    # Caso: dos nombres y un apellido
    elif len(partes) == 3:
        primer_nombre = partes[0]
        primer_apellido = partes[1]
        segundo_apellido = partes[2]

        correo = (
            primer_nombre[0]
            + primer_apellido
            + segundo_apellido[0]
            + "@gmail.com"
        )

    # Caso: dos nombres y dos apellidos
    else:
        primer_nombre = partes[0]
        segundo_nombre = partes[1]
        primer_apellido = partes[2]
        segundo_apellido = partes[3]

        correo = (
            primer_nombre[0]
            + segundo_nombre[0]
            + primer_apellido
            + segundo_apellido[0]
            + "@gmail.com"
        )

    print("Correo generado:")
    print(correo)

except ValueError as e:
    print("Error:", e)
```

```python
import re

try:
    nombre_completo = input("Ingrese su nombre completo: ").lower()

    # Validar que solo haya letras y espacios
    if not re.fullmatch(r"[a-zA-Záéíóúñ\s]+", nombre_completo):
        raise ValueError("Solo se permiten letras y espacios")
    partes = nombre_completo.split()

    # Debe tener al menos nombre y apellido
    if len(partes) < 2:
        raise ValueError("Debe ingresar al menos nombre y apellido")

    # Tomar todos los nombres excepto los últimos dos
    nombres = partes[:-2]

    # Tomar apellidos
    primer_apellido = partes[-2]
    segundo_apellido = partes[-1]

    # Iniciales de todos los nombres
    iniciales = ""

    for nombre in nombres:
        iniciales += nombre[0]

    # Crear correo
    correo = (
        iniciales
        + primer_apellido
        + segundo_apellido[0]
        + "@gmail.com"
    )

    print("Correo generado:")
    print(correo)

except ValueError as e:
    print("Error:", e)
```
