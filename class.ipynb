```python
class Persona:
    def __init__(self, nombre, edad, peso, estatura):
        self.nombre = nombre
        self.edad = edad
        self.peso = peso
        self.estatura = estatura

    def imc(self):
        return self.peso / (self.estatura ** 2)

    def saludar(self):
        print(f"Hola {self.nombre}, tu IMC es {self.imc():.2f}")
```

```python
persona1 = Persona("Alejandra", 21, 87, 1.58)
persona1.saludar()
```

```python
persona2 = Persona("Carlos ", 25, 65, 1.70)
persona2.saludar()
```

```python
class Calculadora:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def sumar(self):
        return self.a + self.b

    def multiplicar(self):
        return self.a * self.b

    def dividir(self):
        try:
            return self.a / self.b
        except ZeroDivisionError:
            return "Error: no se puede dividir por cero"


a = float(input("Ingrese el primer número: "))
b = float(input("Ingrese el segundo número: "))


calc = Calculadora(a, b)

print("Suma:", calc.sumar())
print("Multiplicación:", calc.multiplicar())
print("División:", calc.dividir())
```

```python
class Estudiante:
   """metodo aprovar"""

   def __init__(self, nombre, nota):
        self.nombre = nombre
        self.nota = nota

   def aprobar(self):
        if self.nota >= 3.0:
            print(self.nombre, "aprobó")
        else:
            print(self.nombre, "reprobó")
```

```python
class CuentaBancaria:
    def __init__(self, titular, saldo):
        self.titular = titular
        self.saldo = saldo

    def depositar(self, cantidad):
        if cantidad > 0:
            self.saldo += cantidad

    def retirar(self, cantidad):
        if cantidad > 0 and cantidad <= self.saldo:
            self.saldo -= cantidad

    def mostrar_saldo(self):
        print("Saldo del titular luego del movimineto es :", self.saldo)
```

```python
cuenta = CuentaBancaria("Alejandra", 1000)
cuenta.depositar(500)
cuenta.retrirar(300)
cuenta.mostrar_saldo()
```
