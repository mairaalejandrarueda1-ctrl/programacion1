```python
import numpy as np
import pandas as pd

np.random.seed(123) # para guardar la semilla
```

```python
datos = np.random.normal(loc = 4.0, scale = 0.5, size = 50)

df = pd.DataFrame({
    "nota": datos
})
```

```python
df. describe()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm, gaussian_kde

# Datos (quitamos NaN por si acaso)
data = df["nota"].dropna()

# Parámetros de la normal
mu, std = data.mean(), data.std()

# Rango para las curvas
x = np.linspace(data.min(), data.max(), 500)

# KDE
kde = gaussian_kde(data)
kde_values = kde(x)

# Estilo bonito
sns.set(style="whitegrid", context="talk")

plt.figure(figsize=(10, 6))

# Histograma normalizado
sns.histplot(data, bins=20, stat="density", color="#69b3a2", edgecolor="black", alpha=0.6, label="Histograma")

# Curva normal
plt.plot(x, norm.pdf(x, mu, std), color="#ff6f61", lw=2.5, label=f"Normal ($\\mu$={mu:.2f}, $\\sigma$={std:.2f})")

# KDE
plt.plot(x, kde_values, color="#4c72b0", lw=2.5, linestyle="--", label="KDE (Kernel)")

# Detalles estéticos
plt.title("Distribución de la variable 'nota'", fontsize=16, weight="bold")
plt.xlabel("Nota")
plt.ylabel("Densidad")
plt.legend(frameon=True)
plt.tight_layout()

plt.show()
```

```python
from scipy.stats import shapiro
shapiro(df["nota"])
```

como el valor p es mayor que 0,05, no se rechaza la hipotesis de normalidad con una confianza del 95%


las calificaciones presentan un romedio vercao a 4.0 con una desciacion estandar 0,5

```python
import numpy as np
import pandas as pd

np.random.seed(123)

grupo_A = np.random.normal(loc=4.0, scale=0.4, size=40)
grupo_B = np.random.normal(loc=4.3, scale=0.5, size=40)
```

```python
df = pd.DataFrame({
    "A": grupo_A,
    "B": grupo_B})
```

```python
import
```

```python
import scipy.stats as stats

t_stat, p_val = stats.ttest_ind(df["A"], df["B"])

print("Estadístico t:", t_stat)
print("Valor p:", p_val)
```

como el valor p es mayor que 0,05, se rechaza la hipotesis de normalidad con una confianza del 95%
